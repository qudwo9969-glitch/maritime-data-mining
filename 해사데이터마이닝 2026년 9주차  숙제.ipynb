```
import pandas as pd
from scipy import stats
import os

# 1. 데이터 로드 설정
# 파일이 있으면 읽고, 없으면 테스트용 가상 데이터를 만듭니다.
file_name = 'before_after_worktime.csv'

if os.path.exists(file_name):
    df = pd.read_csv(file_name)
else:
    # 테스트를 위한 임시 데이터 생성 (파일이 없을 때 실행됨)
    data = {
        'before_min': [45, 52, 48, 60, 55, 49, 58, 50, 54, 56],
        'after_min': [40, 48, 45, 55, 50, 48, 53, 47, 51, 52]
    }
    df = pd.DataFrame(data)
    print(f"⚠️ '{file_name}' 파일을 찾을 수 없어 임시 데이터로 분석을 진행합니다.\n")

# 2. 기초 통계량 확인
print('[데이터 미리보기]')
print(df.head(), '\n')

before = df['before_min']
after = df['after_min']
diff = before - after

# 3. 대응표본 t-검정 (Paired Samples t-test)
# H0 (귀무가설): 교육 전후 평균 차이는 없다. (μ_d = 0)
# H1 (대립가설): 교육 후 업무처리시간이 감소했다. (before > after, 즉 μ_d > 0)
result = stats.ttest_rel(before, after, alternative='greater')

# 4. 분석 결과 출력
print('[분석 결과]')
print(f'표본 수: {len(df)}')
print(f'평균 차이(before - after): {round(diff.mean(), 4)}')
print(f't-statistic: {round(result.statistic, 4)}')
print(f'p-value: {round(result.pvalue, 6)}')

# 5. 결과 해석
print('\n[결과 해석]')
alpha = 0.05
if result.pvalue < alpha:
    print(f'p-value가 {alpha}보다 작으므로 귀무가설을 기각합니다.')
    print('통계적으로 유의미하게 교육 후 업무처리시간이 감소했다고 볼 수 있습니다.')
else:
    print(f'p-value가 {alpha}보다 크므로 귀무가설을 채택합니다.')
    print('교육 전후 업무처리시간에 유의미한 차이가 있다고 보기 어렵습니다.')
```

```
import pandas as pd
from scipy import stats
import os

# 1. 데이터 로드 설정
file_name = 'delivery_time.csv'

# 파일이 있으면 읽고, 없으면 테스트용 가상 데이터를 만듭니다.
if os.path.exists(file_name):
    df = pd.read_csv(file_name)
else:
    # 깃허브나 로컬 테스트를 위한 임시 데이터 생성
    data = {
        'delivery_min': [28, 25, 32, 27, 24, 29, 31, 26, 25, 28]
    }
    df = pd.DataFrame(data)
    print(f"⚠️ '{file_name}' 파일을 찾을 수 없어 임시 데이터로 분석을 진행합니다.\n")

# 2. 데이터 확인 및 전처리
# 결측치 제거 (dropna)
x = df['delivery_min'].dropna()

print('[데이터 미리보기]')
print(df.head(), '\n')

# 3. 단일표본 t-test (One-Sample t-test)
# H0 (귀무가설): 평균 배달시간은 30분 이상이다. (μ >= 30)
# H1 (대립가설): 평균 배달시간은 30분 미만이다. (μ < 30)
popmean = 30
result = stats.ttest_1samp(x, popmean=popmean, alternative='less')

# 4. 기초 통계량 및 신뢰구간 계산
mean_x = x.mean()
n = len(x)
se = stats.sem(x)  # 표준오차
# 95% 신뢰구간 계산
ci_low, ci_high = stats.t.interval(0.95, df=n-1, loc=mean_x, scale=se)

# 5. 분석 결과 출력
print('[분석 결과]')
print(f'표본 수: {n}')
print(f'표본평균: {round(mean_x, 4)}')
print(f't-statistic: {round(result.statistic, 4)}')
print(f'p-value: {round(result.pvalue, 6)}')
print(f'95% 신뢰구간: ({round(ci_low, 4)}, {round(ci_high, 4)})')

# 6. 결과 해석
print('\n[결과 해석]')
alpha = 0.05
if result.pvalue < alpha:
    print(f'p-value가 {alpha}보다 작으므로 귀무가설을 기각합니다.')
    print(f'통계적으로 유의미하게 평균 배달시간이 {popmean}분 미만이라고 할 수 있습니다.')
else:
    print(f'p-value가 {alpha}보다 크므로 귀무가설을 채택합니다.')
    print(f'평균 배달시간이 {popmean}분 미만이라고 결론 내릴 통계적 근거가 부족합니다.')
```
